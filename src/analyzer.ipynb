```python
import pandas as pd
import matplotlib.pyplot as plt
import os

CLEANED_DATA_PATH = "../data/cleaned/remoteok_jobs_cleaned.csv"


def load_dataset():
    print("Loading cleaned dataset...")
    df = pd.read_csv(CLEANED_DATA_PATH)
    print(f"Rows loaded: {len(df)}")
    return df


# =============================
#  Visualization 1 — Job Type Distributionimport pandas as pd
import matplotlib.pyplot as plt
import os

CLEANED_DATA_PATH = "../data/cleaned/remoteok_jobs_cleaned.csv"


def load_dataset():
    print("Loading cleaned dataset...")
    df = pd.read_csv(CLEANED_DATA_PATH)
    print(f"Rows loaded: {len(df)}")
    return df


# =============================
#  Visualization 1 — Job Type Distribution
# =============================

def plot_job_type_distribution(df):
    print("Generating job type distribution chart...")

    counts = df["Job Type"].value_counts()

    plt.figure(figsize=(8, 8))
    plt.pie(
        counts,
        labels=counts.index,
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title("Distribution of Job Types in Remote Jobs", fontsize=13)

    os.makedirs("../visualizations", exist_ok=True)
    plt.savefig("../visualizations/job_type_distribution.png", bbox_inches="tight")
    plt.close()

    print("Saved: ../visualizations/job_type_distribution.png")


# =============================
#  Visualization 2 — Top Skills
# =============================

def plot_top_skills(df):
    print("Generating top skills chart...")

    skills_df = df.copy()
    skills_df["Job Tags / Skills"] = skills_df["Job Tags / Skills"].astype(str).str.split(", ")
    skills_df = skills_df.explode("Job Tags / Skills")

    top_skills = skills_df["Job Tags / Skills"].value_counts().head(15)

    plt.figure(figsize=(10, 6))
    top_skills.plot(kind="bar")

    plt.xlabel("Skills / Tags")
    plt.ylabel("Number of Jobs")
    plt.title("Top In-Demand Skills in Remote Job Listings")

    plt.xticks(rotation=45, ha="right")

    os.makedirs("../visualizations", exist_ok=True)
    plt.savefig("../visualizations/top_skills.png", bbox_inches="tight")
    plt.close()

    print("Saved: ../visualizations/top_skills.png")


# =============================
#  Visualization 3 — Top Job Titles
# =============================

def plot_top_job_titles(df):
    print("Generating top job titles chart...")

    top_titles = df["Job Title"].value_counts().head(10)

    plt.figure(figsize=(10, 6))
    top_titles.plot(kind="bar")

    plt.xlabel("Job Title")
    plt.ylabel("Number of Postings")
    plt.title("Top Remote Job Titles")

    plt.xticks(rotation=45, ha="right")

    os.makedirs("../visualizations", exist_ok=True)
    plt.savefig("../visualizations/top_job_titles.png", bbox_inches="tight")
    plt.close()

    print("Saved: ../visualizations/top_job_titles.png")


# =============================
#  Visualization 4 — Skill Frequency Comparison Across Job Types
# =============================

def plot_skill_frequency_comparison(df):
    print("Generating skill frequency comparison chart...")

    skills_df = df.copy()
    skills_df["Job Tags / Skills"] = skills_df["Job Tags / Skills"].astype(str).str.split(", ")
    skills_df = skills_df.explode("Job Tags / Skills")

    pivot = (
        skills_df
        .groupby(["Job Type", "Job Tags / Skills"])
        .size()
        .reset_index(name="Count")
    )

    top = pivot.sort_values("Count", ascending=False).head(15)

    plt.figure(figsize=(12, 6))

    for job_type in top["Job Type"].unique():
        sub = top[top["Job Type"] == job_type]
        plt.plot(sub["Job Tags / Skills"], sub["Count"], marker="o", label=job_type)

    plt.xlabel("Skills / Tags")
    plt.ylabel("Frequency")
    plt.title("Skill Frequency Comparison Across Job Types")

    plt.xticks(rotation=45, ha="right")
    plt.legend()

    os.makedirs("../visualizations", exist_ok=True)
    plt.savefig("../visualizations/skill_frequency_comparison.png", bbox_inches="tight")
    plt.close()

    print("Saved: ../visualizations/skill_frequency_comparison.png")


# =============================
#  Main Runner
# =============================

def main():
    df = load_dataset()

    plot_job_type_distribution(df)
    plot_top_skills(df)
    plot_top_job_titles(df)
    plot_skill_frequency_comparison(df)

    print("All visualizations generated.")


if __name__ == "__main__":
    main()

# =============================

def plot_job_type_distribution(df):
    print("Generating job type distribution chart...")

    counts = df["Job Type"].value_counts()

    plt.figure(figsize=(8, 8))
    plt.pie(
        counts,
        labels=counts.index,
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title("Distribution of Job Types in Remote Jobs", fontsize=13)

    os.makedirs("../visualizations", exist_ok=True)
    plt.savefig("../visualizations/job_type_distribution.png", bbox_inches="tight")
    plt.close()

    print("Saved: ../visualizations/job_type_distribution.png")


# =============================
#  Visualization 2 — Top Skills
# =============================

def plot_top_skills(df):
    print("Generating top skills chart...")

    skills_df = df.cop_
```
